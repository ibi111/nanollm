```python
import regex as re
```

```python
pattern = r"""(?i:'s|'t|'re|'ve|'m|'ll|'d)|[^\r\n\p{L}\p{N}]?[\p{L}\p{M}]+|\p{N}| ?[^\s\p{L}\p{M}\p{N}]+[\r\n]*|\s*[\r\n]+|\s+(?!\S)|\s+""" # qwen
```

```python
pattern = r"""'(?i:[sdmt]|ll|ve|re)|[^\r\n\p{L}\p{N}]?+\p{L}+|\p{N}{1,3}| ?[^\s\p{L}\p{N}]++[\r\n]*|\s*[\r\n]|\s+(?!\S)|\s+""" # gpt splits numbers as a pair of 3 hence providing better compression at math or numbers majority corpora
```

```python
compiled_pattern = re.compile(pattern)
```

```python
text = "Hello, world! 123\n"
pieces = compiled_pattern.findall(text)

print("Original:", repr(text))
print("Reconstructed:", repr("".join(pieces)))
print("Preserved:", text == "".join(pieces))
```
